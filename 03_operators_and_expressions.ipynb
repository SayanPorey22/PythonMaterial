## 3. Operators and Expressions

*Scope:* Operator categories and how expressions are evaluated.

### 3.1 Operator Categories

Python's operators fall into 8 categories — each covered in its own sub-section below,
with the operator symbols and a simple example.

* 3.1.1 Arithmetic Operators
* 3.1.2 Relational Operators
* 3.1.3 Boolean Operators
* 3.1.4 Non-Boolean Operators
* 3.1.5 Bitwise Operators
* 3.1.6 Assignment Operators
* 3.1.7 Ternary Operators
* 3.1.8 Special Operators — Identity and Membership

#### 3.1.1 Arithmetic Operators

Perform basic math: `+`, `-`, `*`, `/`, `//` (floor division), `%` (modulus), `**`
(exponent).

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 3)   # 3.5 3 1 8
```

#### 3.1.2 Relational Operators

Compare two values and always produce a `bool`: `==`, `!=`, `>`, `<`, `>=`, `<=`.

```python
print(5 > 3, 5 == 3)   # True False
```

#### 3.1.3 Boolean Operators

Combine `bool` expressions: `and`, `or`, `not`. With `bool` operands the result is always
a `bool`.

```python
print(True and False, True or False, not True)   # False True False
```

#### 3.1.4 Non-Boolean Operators

The same `and` / `or` operators also work on **non-`bool`** operands — they short-circuit
based on truthiness and return one of the actual **operands**, not `True`/`False`.

```python
print(3 and 5)   # 5   -> both truthy, `and` returns the last one
print(0 and 5)   # 0   -> first is falsy, `and` short-circuits and returns it
print(3 or 5)    # 3   -> first is truthy, `or` short-circuits and returns it
```

#### 3.1.5 Bitwise Operators

Operate on the binary representation of integers: `&`, `|`, `^`, `~`, `<<`, `>>`.

```python
print(5 & 3, 5 | 3)   # 1 7   -> 0b101 & 0b011 = 0b001, 0b101 | 0b011 = 0b111
```

#### 3.1.6 Assignment Operators

**Basic assignment (`=`)** binds a name to an object. It supports:

* **Chained assignment** — `a = b = c = 5` binds all three names to the *same* object.
* **Multiple assignment (unpacking)** — `a, b = 1, 2` assigns element-wise from an
  iterable.

**Augmented assignment** (`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`, `&=`, `|=`, `^=`,
`<<=`, `>>=`) updates a variable using its current value in one step. For **immutable**
types (`int`, `str`, `tuple`, ...) it behaves exactly like `x = x + 3` — a new object is
created and `x` is rebound to it. For **mutable** types (`list`, `dict`, `set`, ...) it can
instead mutate the *same* object in place, if the type defines the matching in-place method
(e.g. `list.__iadd__`).

```python
x = [1, 2]
y = x
x += [3]           # in-place: mutates the same object
print(y)            # [1, 2, 3] -> y sees the change, it's still the same object

x = [1, 2]
y = x
x = x + [3]        # NOT in-place: builds a new list, x rebinds to it
print(y)            # [1, 2]    -> y is unaffected, it still points to the old object
```

#### 3.1.7 Ternary Operators

Called *ternary* because it takes three operands — a condition, a true-value and a
false-value: `value_if_true if condition else value_if_false`.

**It's an expression, not a statement** — since it evaluates to a value, it can be used
anywhere a value is expected: in an assignment, as a function argument, inside a
comprehension, or inside an f-string. A full `if`/`else` *statement* cannot be used in any
of those places.

```python
x = 7
result = "even" if x % 2 == 0 else "odd"                  # assignment
print(f"{x} is {'even' if x % 2 == 0 else 'odd'}")          # inside an f-string
```

**Chaining** — ternaries can be nested for more than two outcomes, though readability
suffers quickly:

```python
x = 0
print("positive" if x > 0 else "negative" if x < 0 else "zero")   # zero
```

**No dedicated `?:` syntax** — unlike C / Java / JavaScript's `condition ? a : b`, Python
spells the same idea out with the `if` / `else` keywords instead of punctuation.

#### 3.1.8 Special Operators — Identity and Membership

**Identity** — `is`, `is not` — compare object **identity** (`id()`), not value.

```python
x = [1, 2, 3]
y = [1, 2, 3]
print(x == y, x is y)   # True False -> equal value, different objects
```

**Membership** — `in`, `not in` — check whether a value exists within a sequence or
collection.

```python
print(3 in [1, 2, 3])   # True
```

### 3.2 Precedence and Associativity

**Precedence** decides which operator binds first when an expression mixes several of
them. From highest to lowest:

| # | Operators | Category |
|---|---|---|
| 1 | `()` | Parentheses (grouping) |
| 2 | `**` | Exponentiation |
| 3 | `+x`, `-x`, `~x` | Unary plus / minus / bitwise not |
| 4 | `*`, `/`, `//`, `%` | Multiplication, division, floor division, modulus |
| 5 | `+`, `-` | Addition, subtraction |
| 6 | `<<`, `>>` | Bitwise shifts |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `\|` | Bitwise OR |
| 10 | `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `is not`, `in`, `not in` | Comparisons |
| 11 | `not` | Boolean NOT |
| 12 | `and` | Boolean AND |
| 13 | `or` | Boolean OR |
| 14 | `x if c else y` | Ternary |
| 15 | `=`, `+=`, `-=`, ... | Assignment |

```python
print(2 + 3 * 4)   # 14 -> * (row 4) binds tighter than + (row 5)
```

**Associativity** decides the order when operators of the *same* precedence appear
together. Most operators are **left-to-right**; `**` is the one exception — it is
**right-to-left**:

```python
print(2 ** 3 ** 2)   # 512 -> evaluated as 2 ** (3 ** 2) = 2 ** 9, not (2 ** 3) ** 2 = 64
```

Assignment is also right-to-left, which is what makes chained assignment (3.1.6) work:
`a = b = 5` assigns to `b` first, then assigns that same object to `a`.

**Parentheses always win** — regardless of precedence or associativity, `()` can force any
part of an expression to evaluate first:

```python
print((2 + 3) * 4)   # 20 -> parentheses override the default precedence
```
