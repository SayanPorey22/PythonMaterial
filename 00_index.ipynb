## 0. How to Use This Notebook

This chapter is the only part of the notebook that is *not* filled from the handwritten
notes. It documents the conventions that keep the structure stable while the content grows.

### 0.1 Purpose and Scope

This notebook is intended to become a single, complete Python reference built from a
detailed set of handwritten notes.

* The chapter list (top level) is meant to stay **stable**.
* Sub-sections are meant to be **split, renamed and deepened** as notes arrive.
* Nothing here is a syllabus to be followed — it is a set of containers to be filled.

### 0.2 Source of Truth

The **handwritten notes are authoritative** for:

* which concepts exist,
* how deep each concept goes,
* which examples, experiments, edge cases and observations are recorded,
* the wording of definitions where the notes are explicit.

Where a section here has no matching note yet, it stays as a placeholder rather than being
filled with generic material.

### 0.3 Per-Concept Documentation Template

Full style guide: [`CLAUDE.md`](CLAUDE.md). Summary — every concept is
**explanation + runnable demo, never one alone**:

1. A markdown cell with a real explanation — not a one-liner. Use whatever fits best: a
   small table (operators, types, syntax forms), a short "why it matters" note, a named
   callout for a nuance or gotcha (`**Common mistake** — ...`), or a contrast with a
   related/confusable concept.
2. One or more **actual code cells** (not fenced ` ```python ` blocks inside markdown),
   each minimal and focused on one thing — a couple of `print()` calls with an inline `#`
   comment showing the expected output.
3. A concept-specific gotcha or a second angle worth showing gets its own small
   markdown note + its own small code cell, rather than one giant cell for everything.

Make abstract ideas (identity, mutability, the object model) tangible — e.g. print
`id()` side by side instead of only describing it in prose.

### 0.4 Cell Rhythm for Each Concept

The cell sequence when a concept is written up:

```text
Markdown  ->  explanation (table / prose / "why it matters")
Code      ->  minimal runnable demo, inline # comments show expected output
Markdown  ->  (optional) a named gotcha or second angle
Code      ->  (optional) its own small demo cell
```

Repeat per concept/sub-concept. Each chapter ends with a scratch cell for ad-hoc
experiments — promote anything worth keeping into a proper section instead of leaving it
there.

### 0.5 Numbering and Insertion Rules

Conventions that avoid renumbering the whole notebook every time a note is added:

1. **Chapter numbers are frozen.** New major domains are appended, never inserted.
2. **New sub-sections are appended** to the end of a chapter, before the `x.9` bucket
   (if the chapter has one — see rule 3).
3. **`x.9 Additional ... From Notes`** is an *optional* per-chapter overflow bucket.
   Anything with no obvious home can land here first and migrate once a pattern
   emerges — but a chapter that is otherwise complete does not need to keep an empty
   one around; it can be dropped.
4. **Deeper levels are free-form.** `13.5.2`, `13.5.3` etc. can be created and reordered
   without touching anything above them.
5. **Moving, renaming or removing a section** means updating the cells and fixing any
   pointer that referenced the old location — in other chapters and in the Appendix
   (§24.2, §24.3) — rather than leaving it dangling.

### 0.6 Status Tags

**Deprecated — not used from chapter 2 onward.** Section headings stay clean (no
`[DONE]`/`[EMPTY]`/...); use §24.2's Chapter Coverage Tracker instead to track what's
filled vs. outstanding. Kept here only as a historical note — chapter 1 briefly used:

* `[EMPTY]` — nothing transferred yet
* `[PARTIAL]` — some notes transferred, more exist on paper
* `[DONE]` — notes fully transferred
* `[VERIFY]` — transferred but the behaviour has not been re-tested in code
* `[EXPAND]` — needs more examples or experiments than the notes contain

### 0.7 Incremental Workflow

```text
Phase 1   Build the skeleton                        <- current state
Phase 2   Transfer notes for one domain at a time
Phase 3   Expand that chapter's sub-sections
Phase 4   Route small / scattered concepts (see 21 and 23)
Phase 5   Repeat until the handwritten notes are exhausted
```

Small concepts arriving out of order go to **23. Inbox** and are filed later. Unrelated
chapters are never restructured to accommodate a single new concept.

### 0.8 File Layout

The notebook was originally a single file; it is now split for size and editability:

* `00_index.ipynb` (this file) — Chapter 0 conventions, Chapter 23 Inbox, Chapter 24 Appendix.
* `NN_<slug>.ipynb` — one notebook per chapter (1–22), e.g. `01_python_fundamentals.ipynb`.

Chapter numbering, the per-concept template, status tags and the `.9` overflow-bucket convention are unchanged — they now just apply within each chapter's own file. In-notebook `see also` pointers (e.g. "see 2.6") refer to a section number, not a file path; resolve them by chapter number to the matching `NN_*.ipynb`.
