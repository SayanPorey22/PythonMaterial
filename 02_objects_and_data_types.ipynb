## 2. Objects and Data Types

*Scope:* What a Python object is, and the built-in type categories with their defining properties.

### 2.1 Overview of Built-in Data Types

**Object** — every value in Python is an object. Every object has an **identity**
(`id()`), a **type** (`type()`) and a **value**; each data type below is simply a built-in
category of object, with its own value representation and behaviour.

**Basic Data Types**

| Type | Definition | Declaration |
|---|---|---|
| `int` | Whole numbers of arbitrary precision. | `x = 10` |
| `float` | Decimal / floating-point numbers. | `x = 10.5` |
| `complex` | Numbers with a real and an imaginary part. | `x = 2 + 3j` |
| `bool` | Boolean truth value — `True` or `False`. | `x = True` |
| `str` | Sequence of Unicode characters (text). | `x = "hello"` |
| `NoneType` | Represents the absence of a value; its only instance is `None`. | `x = None` |

**Composite Data Types**

| Type | Definition | Declaration |
|---|---|---|
| `list` | Ordered, mutable collection of items. | `x = [1, 2, 3]` |
| `tuple` | Ordered, immutable collection of items. | `x = (1, 2, 3)` |
| `set` | Unordered, mutable collection of unique items. | `x = {1, 2, 3}` |
| `frozenset` | Unordered, immutable collection of unique items. | `x = frozenset({1, 2, 3})` |
| `dict` | Collection of key-value pairs (insertion-ordered). | `x = {"a": 1, "b": 2}` |
| `bytes` | Immutable sequence of raw bytes. | `x = b"hello"` |
| `bytearray` | Mutable sequence of raw bytes. | `x = bytearray(b"hello")` |
| `range` | Immutable, lazily generated sequence of numbers. | `x = range(10)` |

### 2.2 Basic Data Types

These are the 5 basic data types in Python — `int`, `float`, `bool`, `str` and `complex`.
All five are **immutable**: once created, the object itself can never be changed — a name
bound to one always points to that single object, and anything that looks like a
modification actually creates a new object and rebinds the name to it.

Deep dive into each one below — literal forms, conversions and inbuilt behaviour, one
sub-section per type.

#### 2.2.1 Integers (`int`)

Python has three numeric types — `int`, `float` and `complex` (defined in 2.1). Of the
three, `int` is the only one with multiple literal forms, since a whole number can be
written in decimal, binary, octal or hexadecimal and still be the exact same value/object.

**Integer literals in different bases**

| Base | Prefix | Example | Value |
|---|---|---|---|
| Decimal (base 10) | none | `x = 42` | 42 |
| Binary (base 2) | `0b` / `0B` | `x = 0b101010` | 42 |
| Octal (base 8) | `0o` / `0O` | `x = 0o52` | 42 |
| Hexadecimal (base 16) | `0x` / `0X` | `x = 0x2A` | 42 |

The prefix only affects how the literal is *written* — all four lines above produce the
same `int` object (`type(x)` is `int` in every case; the base is not stored anywhere).

**Converting between bases**

* `int` → base-prefixed string: `bin()`, `oct()`, `hex()`

```python
bin(42)   # '0b101010'
oct(42)   # '0o52'
hex(42)   # '0x2a'
```

* string in a given base → `int`: `int(string, base)`

```python
int('101010', 2)   # 42
int('52', 8)        # 42
int('2a', 16)       # 42
```

#### 2.2.2 Floating-Point Numbers (`float`)

A number with a decimal point (or written in exponential notation) is a `float`.

```python
x = 10.5
y = 1.5e3      # exponential notation -> 1500.0
```

#### 2.2.3 Booleans (`bool`)

A `bool` holds one of two truth values, `True` or `False`.

```python
x = True
```

#### 2.2.4 Strings (`str`)

A `str` is a sequence of characters, written inside single, double or triple quotes.

```python
x = "hello"
```

If the string holds a valid whole number, it can be typecast to `int` with `int()`:

```python
int("42")   # 42
```

#### 2.2.5 Complex Numbers (`complex`)

A `complex` number has a real part and an imaginary part, the latter suffixed with `j`.

```python
x = 2 + 3j
```

### 2.3 Composite Data Types

These are the 8 composite data types in Python — `list`, `tuple`, `set`, `frozenset`,
`dict`, `bytes`, `bytearray` and `range`. Unlike the basic types (2.2), mutability varies
by type: `list`, `set`, `dict` and `bytearray` are **mutable**; `tuple`, `frozenset`,
`bytes` and `range` are **immutable**.

Deep dive into each one below — literal forms and inbuilt behaviour, one sub-section per
type.

#### 2.3.1 Lists (`list`)

An ordered, mutable collection of items.

```python
x = [1, 2, 3]
```

**Mutable** means the object can be changed in place after creation — it stays the *same*
object (same `id()`), just with different contents:

```python
x = [1, 2, 3]
print(id(x))      # e.g. 140023456
x.append(4)        # changes the object in place
print(x, id(x))     # [1, 2, 3, 4]  140023456  <- same id, same object
```

#### 2.3.2 Tuples (`tuple`)

An ordered, immutable collection of items.

```python
x = (1, 2, 3)
```

**Ordered** means each item keeps a fixed position, accessible by index:

```python
x = (10, 20, 30)
print(x[0], x[1], x[2])   # 10 20 30 -> items keep their position/index
```

**Immutable** means the object can never be changed in place after creation — anything
that looks like a modification actually builds a *new* object:

```python
x = (1, 2, 3)
print(id(x))       # e.g. 140099876
x = x + (4,)         # no in-place way to add an item -> builds a new tuple
print(x, id(x))     # (1, 2, 3, 4)  <- different id, a different object
```

#### 2.3.3 Sets (`set`)

An unordered, mutable collection of unique items.

```python
x = {1, 2, 3}
```

#### 2.3.4 Frozensets (`frozenset`)

An unordered, immutable collection of unique items.

```python
x = frozenset({1, 2, 3})
```

#### 2.3.5 Dictionaries (`dict`)

A collection of key-value pairs, insertion-ordered.

```python
x = {"a": 1, "b": 2}
```

It can also be created with the `dict()` constructor:

```python
x = dict(a=1, b=2)   # {'a': 1, 'b': 2}
```

Being mutable, a new key-value pair can be added in place:

```python
x = {"a": 1, "b": 2}
x["c"] = 3
print(x)   # {'a': 1, 'b': 2, 'c': 3}
```

#### 2.3.6 Bytes (`bytes`)

An immutable sequence of raw bytes.

```python
x = b"hello"
```

#### 2.3.7 Bytearrays (`bytearray`)

A mutable sequence of raw bytes.

```python
x = bytearray(b"hello")
```

Each element is a single byte — an integer in the range `0` to `255` (a byte is 8 bits,
i.e. 2⁸ = 256 possible values). Assigning a value outside this range raises `ValueError`.

#### 2.3.8 Ranges (`range`)

An immutable, lazily generated sequence of numbers.

```python
x = range(10)
```

### 2.4 Object Reuse and Caching (Interning)

**How reassignment works**

```python
x = 11
x = 10
```

`x` is first bound to an `int` object `11`. The second line does not change that object —
it creates a **new** `int` object `10` and rebinds `x` to it. If nothing else refers to the
`11` object anymore, it becomes eligible for garbage collection.

```python
y = 10
```

Now `y` is bound to the same value `10`. CPython checks whether an object for that value
already exists in a small internal cache and, if so, reuses it instead of creating a new
one — so `y` ends up pointing at the **same object** as `x`, and `id(x) == id(y)`.

This is a **CPython implementation detail**, not a language guarantee — it exists purely as
a memory/speed optimization for common, immutable values. Never rely on `is` for value
comparison; use `==`.

**Caching ranges by type**

| Type | Cached / interned? | Range / rule |
|---|---|---|
| `int` | Yes | Small integers from `-5` to `256` (inclusive) are pre-created and reused |
| `bool` | Always | Only two `bool` objects ever exist — the singletons `True` and `False` |
| `str` | Sometimes | Identifier-like literals (letters/digits/underscore only) and compile-time constants are interned; arbitrary or computed strings usually are not |
| `float` | No | Every `float` object is distinct, even for the same value |
| `complex` | No | Every `complex` object is distinct, even for the same value |

```python
# int - cached for -5 to 256
a, b = 100, 100
print(a is b)      # True  -> same cached object

a, b = 300, 300
print(a is b)      # False -> outside the cached range, separate objects
```

```python
# bool - always the same singleton
a, b = True, True
print(a is b)      # True
```

```python
# str - interned only for identifier-like literals
a, b = "hello", "hello"
print(a is b)      # True  -> identifier-like, interned

a, b = "hello world!", "hello world!"
print(a is b)      # False -> contains space/punctuation, not interned
```

```python
# float - never cached
a, b = 10.5, 10.5
print(a is b)      # False -> separate objects every time
```

```python
# complex - never cached
a, b = 2 + 3j, 2 + 3j
print(a is b)      # False -> separate objects every time
```

**Composite (mutable) types never share objects** — `list`, `set`, `dict`, `bytearray`,
etc. always create a brand-new object on every assignment, regardless of value. Caching
only makes sense for values that can never change; sharing a *mutable* object across
unrelated names would let a change made through one name silently affect the other.

```python
a, b = [1, 2, 3], [1, 2, 3]
print(a is b)      # False -> always a new object
```
